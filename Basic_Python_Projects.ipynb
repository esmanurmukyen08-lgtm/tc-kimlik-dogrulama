1) Depo Stok Kontrol Sistemi

Bir depoda:

laptop = 12
mouse = 25
klavye = 10

Kullanıcı ürün adı ve kaç adet istediğini girsin.

Kurallar:

Stok yeterliyse “satış başarılı”
Yetmiyorsa “stok yetersiz”
yanlış ürün girerse hata versin
kullanıcı "çıkış" yazana kadar devam etsin

```python
laptop=12
mouse=25
klavye=10

while True:
  ürün=input("ürün adını giriniz:(laptop/mouse/klavye/çıkış)")

  if ürün=="çıkış":
    print("sistem kapatılıyor")
    break

  adet=int(input("kaç adet alıcaksınız:"))

  if ürün=="laptop":
    if adet<=laptop:
      print("satış başarılı")
      laptop -= adet
    else:
      print("yetersiz stok")

  elif ürün=="mouse":
    if adet<=mouse:
      print("satış başarılı")
      mouse -= adet
    else:
      print("yetersiz stok")

  elif ürün=="klavye":
    if adet<=klavye:
      print("satış başarılı")
      klavye -= adet
    else:
      print("yetersiz stok")

  else:
    print("yanlış ürün girdiniz")
```

2) Banka ATM Sistemi

Başlangıç bakiyesi: 5000 TL

Kullanıcı:

para çekme
para yatırma
bakiye görüntüleme

işlemleri yapabilsin.

Kurallar:

bakiyeden fazla para çekilemez
negatif sayı girilirse hata ver
çıkış yapılana kadar sistem dönsün

```python
başlangıç_bakiyesi=5000

while True:
  print("\n-----atm sistemi----")
  print("1-para çekme")
  print("2-para yatırma")
  print("3-bakiye görüntüleme")
  print("4-çıkış")

  seçim=input("seçiminizi yapınız:")

  if seçim=="1":
    çekilecek_tutar=int(input("çekmek istediğiniz tutarı giriniz:"))
    if çekilecek_tutar>5000 and çekilecek_tutar<0:
      print("yetersiz bakiye")
    else:
      print(f"kalan paranız:{başlangıç_bakiyesi-çekilecek_tutar}")

  elif seçim=="2":
    yatıralacak_para=int(input("yatırmak istediğiniz tutarı giriniz:"))
    print(f"yeni bakiyeniz:{başlangıç_bakiyesi+yatıralacak_para}")

  elif seçim=="3":
    print(f"bakiyeniz:{başlangıç_bakiyesi}")

  elif seçim=="4":
    print("sistem kapanıyor")
    break

```

3) Hastane Sıra Sistemi

Liste:

sira = []

İşlemler:

hasta ekleme
hasta çağırma
sırayı görüntüleme

Kurallar:

İlk gelen ilk çağrılsın
sıra boşsa uyarı versin
sistem sürekli çalışsın

```python
liste=[]
while True:
  print("\n----hastane sıra sitemi----")
  print("1-hasta ekleme")
  print("2-hasta çağırma")
  print("3-sırayı görüntüleme")
  print("4-çıkış")

  seçim=input("seçim yapınız:")

  if seçim=="1":
    hasta=input("hastanın adı:")
    liste.append(hasta)
    print(f"{hasta} sıraya eklendi")

  elif seçim=="2":
    if len(liste)==0:
      print("sırada hasta yok")
    else:
      çağrılan=liste.pop(0)
      print(f"sıradaki hasta:{çağrılan}")

  elif seçim=="3":
    if len(liste)==0:
      print("bekleyen hasta yok")
    else:
      print("Bekleyen hastalar:")
      for i in liste:
        print(f"- {i}")

  elif seçim=="4":
    print("sistem kapanıyor")
    break

  else:
    print("yanlış seçim yaptınız")
```

5) Market İndirim Sistemi

Kullanıcı alışveriş tutarı girsin.

Kurallar:

1000 altı → indirim yok
1000-3000 → %10
3000 üstü → %20

Ek olarak:

öğrenci ise ekstra %5 indirim



```python
alışveriş_tutarı=int(input("alışveriş tutarını giriniz:"))
öğrenci_mi=input("öğrenci misiniz?(evet/hayır)")

if öğrenci_mi=="evet":
  print(f"indirimli fiyatınız:{alışveriş_tutarı-(alışveriş_tutarı*5/100)}")
else:
  print("öğrenci indirimi uygulanmadı.")

# Genel indirim kuralları
if alışveriş_tutarı<1000:
  print("Ek indirim yok.")
elif 1000<=alışveriş_tutarı<3000:
  print(f"Genel %10 indirimli fiyat:{alışveriş_tutarı-alışveriş_tutarı*10/100}")
elif alışveriş_tutarı>=3000:
  print(f"Genel %20 indirimli fiyatınız:{alışveriş_tutarı-alışveriş_tutarı*20/100}")
```

7) Şirket Giriş Sistemi

Doğru bilgiler:

kullanici = "admin"
sifre = "1234"

Kullanıcı doğru girene kadar tekrar istensin.

Kurallar:

doğruysa giriş başarılı
yanlışsa tekrar dene
3 yanlışta sistem kilitlensin


```python
kullanıcı="admin"
şifre="1234"
hak=3
while hak>0:
  kullanıcı_girdisi=input("kullanıcı ismini giriniz:")
  şifre_girdisi=int(input("şifrenizi giriniz:"))
  if kullanıcı==kullanıcı_girdisi and şifre==şifre_girdisi:
    print("giriş yapıldı")
  else:
    hak-=1
    print(f"kalan hak:{hak}")

if hak==0:
  print("sistem kapandı tekrar deneyiniz")
```

9) Fabrika Kalite Kontrol Sistemi

Ürün puanı girilecek.

Kurallar:

90+ → A kalite
70-89 → B kalite
50-69 → C kalite
50 altı → hatalı ürün

Ek:

kullanıcı sürekli ürün puanı girebilsin
çıkış için -1 girsin


```python
while True:
  puan=int(input("ürün puanını giriniz:(çıkış için =-1 basınız)"))

  if puan==-1:
    print("sistem kapatılıyor")
    break

  if puan>=90:
    print("A kalite")
  elif 70<=puan<=89:
    print("B kalite")
  elif 50<=puan<=69:
    print("C kalite")
  elif puan<50:
    print("hatalı ürün")
```
